**Вопрос 6.**

Заполним пропущенные значения.

```python
#nanmedian - считает медианное значение
#не учитывая пропуски
np.nanmedian(data['horsepower']) #сейчас значение 254.0

#найдем значение, которое встречается чаще всего
data['horsepower'].value_counts()

#заполним пропуски самым частым значением 252.0
data['horsepower'] = data['horsepower'].fillna(252)

#узнаем изменилось ли оно
np.median(data['horsepower'])
#оно уменьшилось
```

**Вопрос 7.**
Найдем веса линейной модели.

```python
#только из Азии
mask = data['origin'] == 'Asia'
#создадим y
y = np.array([1100, 1300, 800, 900, 1000, 1100, 1200])
#создадим x
x = data.loc[mask, ['vehicle_weight', 'model_year']].iloc[:7]
x.head(7)
```

Проведем вычисление весов.

```python
x = x.to_numpy()

xtx = x.T @ x
xtx_inv = np.linalg.inv(xtx)
w = xtx_inv @ x.T @ y
print(sum(w))
```
