### **Посмотрим на встроенный метод отбора признаков. И далее посмотрим на методы обертки.**

```python
import pandas as pd
```

_Считаем данные._

```python
x = pd.read_csv('../5. Обобщающая способность/Практика с кик старта/X.csv').drop('Название', axis=1)
y = pd.read_csv('../5. Обобщающая способность/Практика с кик старта/Y.csv')
x.head()
```

_Используем лассо-регрессию. Чем больше значение alpha, тем больше признаков исчезнет._

```python
from sklearn.linear_model import Lasso
#обучим
model = Lasso(alpha=15)
model.fit(x, y)
#увидим, что некоторые признаки обнулились
model.coef_
```

### **Методы обертки.**

_Метод прямого отбора. Сначала берем один лучший признак, потом другой._

```python
x.shape
```

```python
from sklearn.linear_model import LinearRegression
model = LinearRegression()

from sklearn.feature_selection import SequentialFeatureSelector
#direction='forward' - прямой
sfs = SequentialFeatureSelector(model, n_features_to_select=5, direction='forward')
sfs.fit(x, y)
```

_Узнаем, какие признаки мы взяли, а какие нет._

```python
sfs.get_support() #булевый список
sfs.get_feature_names_out() #названия
x[sfs.get_feature_names_out()]
```

_Метод обратного отбора. Смотрим, какой один признак выбросить._

```python
model = LinearRegression()

from sklearn.feature_selection import SequentialFeatureSelector
#direction='backward' - обратный
sfs = SequentialFeatureSelector(model, n_features_to_select=30, direction='backward')
sfs.fit(x, y)
```

_Увидим эти признаки._

```python
sfs.get_support() #булевый список
sfs.get_feature_names_out() #названия
x[sfs.get_feature_names_out()]
```
