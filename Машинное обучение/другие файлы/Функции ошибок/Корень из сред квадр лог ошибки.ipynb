```python
import numpy as np
y = np.array([10, 15, 20])
x_under = np.array([8, 10, 14])
x_over = np.array([12, 20, 26])
rmse_over = np.mean((x_over - y) ** 2) ** 0.5
rmse_under = np.mean((x_under - y) ** 2) ** 0.5
```

```python
rmse_over, rmse_under
```

```python
y_log = np.log(y + 1)
x_log_over = np.log(x_over + 1)
x_log_under = np.log(x_under + 1)
```

```python
rmsle_over = (((x_log_over - y_log)**2).mean())**0.5
rmsle_under = (((x_log_under - y_log)**2).mean())**0.5
rmsle_under, rmsle_over
```
