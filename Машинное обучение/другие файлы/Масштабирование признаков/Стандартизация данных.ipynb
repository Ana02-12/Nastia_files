```python
import numpy as np
import pandas as pd
```

```python
features = np.array([1, 4, 50, 30, 10_000])
```

__вычислим стандартное отклонение__

```python
average = features.mean() #среднее знач признака
deviations = features - average #отличие каждого знач от среднего
squared = deviations ** 2 #квадрат, чтоб избавится от знака
variance = squared.mean() #дисперсия (среднее квадратов отклонений)
std = np.sqrt(variance) #стандартное отклонение
#или
std_ = features.std()
std, std_
```

**стандартизируем признаки**

_1. Вычитаем среднее -> центрируем данные относительно 0._

_2. Делим на стандартное отклонение -> приводим разброс к единому масштабу._

```python
features_std = (features - average) / std  #кол-во стандартных отклонений от среднего
features_std
```
