```python
import numpy as np
import pandas as pd
import seaborn as sns
```

Данные.

```python
featers = np.array([
    [150, 40],
    [140, 35],
    [155, 45],
    [185, 95],
    [145, 40],
    [195, 100],
    [180, 95],
    [170, 80],
    [160, 90]
])
```

Начальные веса.

```python
weights = np.array([0.2, 0.3])
```

Ответы.

```python
true_predictions = np.array([0, 0, 0, 100, 0, 100, 100, 100, 100])
```

Полнопакетный градиентный спуск.

```python
lr = 0.000001
def pack_grade(feat, true, weights, lr):
    pred = feat @ weights.T #вектор (9,0)
    error = true - pred #вектор (9,0)
    #найдем градиент для каждого веса
    grads = []
    #нам нужно вычислить градиент двух его признаков
    #для каждого объекта
    for i in range(len(pred)):
        grads_for_object = []
        #для каждого признака будет свой градиент
        for j in range(len(weights)):
            grads_for_object.append(error[i] * featers[i][j])
        grads.append(grads_for_object)
    #теперь посчитаем средний градиент для каждого веса
    grads = np.array(grads).mean(axis=0)
    #теперь применим его к нашим весам
    weights += grads * lr
    return weights
```

```python
for _ in range(1000):
    weights = pack_grade(featers, true_predictions, weights, lr)
    pred = featers @ weights.T
    print(((true_predictions - pred)**2).mean())
```
