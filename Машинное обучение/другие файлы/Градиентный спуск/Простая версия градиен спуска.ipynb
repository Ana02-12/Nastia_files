```python
import numpy as np
import seaborn as sns
```

```python
price = np.array([4, 5, 6.5])
rooms = np.array([1, 2, 3])
b = 1
k = -2
lr = 0.05
x = np.array([1, 2, 3])
y = x * k + b
sns.scatterplot(x = rooms, y=price)
sns.lineplot(x = rooms, y = price)
sns.lineplot(x = x, y = y, color='#f00')
```

```python
def f(b, k, rooms, price, lr):
    pred = rooms * k + b
    b += (price - pred).mean() * lr
    k += ((price - pred) * rooms).mean() * lr
    return b, k
```

```python

b, k = f(b, k, rooms, price, lr)
y = x * k + b
sns.scatterplot(x = rooms, y=price)
sns.lineplot(x = rooms, y = price)
sns.lineplot(x = x, y = y, color='#f00')
```
