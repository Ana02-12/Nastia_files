```python
import pandas as pd
import numpy as np
import seaborn as sns
```

Имеем один признак и таргет к нему

```python
feater = np.array([-2, 5, 10, 12])
target = np.array([-20, 50, 100, 120])
```

Изобразим эти точки и прямую, которая будет их предсказывать

```python
sns.scatterplot(x=feater, y=target, color='darkviolet')
k, b = 0, 0
sns.lineplot(x=feater, y=feater * k + b, color='green')
```


Функция квадратического подхода.
Внутри будет изменять положение прямой.

```python
def squar_tric(k, b, feat, tar, pred, learn):
    #подсчет градиента
    grad_k = np.mean((pred - tar) * feat)
    grad_b = np.mean(pred - tar)
    #пересчет коэффициентов
    k -= learn * grad_k
    b -= learn * grad_b
    return k, b
```

```python
sns.scatterplot(x=feater, y=target, color='darkviolet')
for _ in range(10):
    sns.lineplot(x=feater, y=k * feater + b, color='green', alpha=0.2)
    #подсчет предсказания
    pred = k * feater + b
    #новые коэффициенты
    k, b = squar_tric(k, b, feater, target, pred, 0.01)
```
