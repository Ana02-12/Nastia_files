```python
import numpy as np
```

_Создадим признаки._

```python
X1 = np.array([20, 24, 35, 52])
X2 = np.array([120, 140, 150, 160])
X3 = np.array([1, 1, 1, 1])
```

_Обозначим коэффициенты, которые хотим найти._

```python
Y = 7 * X1 + 5 * X2 + 2 * X3 #вектор (4,)
Y
```

_Обозначим таблицу с данными._

```python
X = np.column_stack([X1, X2])
# X = np.stack([X1,X2], axis=1)
# X = np.c_[X1, X2]
X
```

_Создадим случайные веса._

```python
w = np.zeros(X.shape[1] + 1)
w
```

### **Функция полно пакетного градиентного спуска.**

```python
lr = 0.01
def grad(X, w, Y, lr):
    #посчитаем предсказание для каждого объекта
    pred = X @ w
    #найдем ошибку по каждому объекту
    error = (pred - Y)[:, None]
    #найдем средний градиент для каждого признака
    grads = 2 * (X * error).mean(axis=0)
    #изменим вес в соответствии с градиентом
    w = w - grads * lr
    return w
```

_Коэффициент b обучается намного медленнее остальных, так как разница между признаками Х1 Х2 и Х3 значительна. Для этого мы приведем все наши данные кроме b к одинаковым единицам измерения - количество стандартных отклонений от среднего значения._

### **Функция для стандартизации данных**

```python
def standart(features):
    #посчитаем средние значения
    f_means = features.mean(axis=0) #вектор
    #найдем квадраты отклонения от среднего
    f_deviation = (features - f_means)**2 #матрица
    #найдем среднее квадратов отклонений - дисперсия
    f_variance = f_deviation.mean(axis=0) #вектор
    #вычислим стандартное отклонение
    f_std = f_variance**0.5 #вектор
    #стандартизируем признаки
    f_stand = (features - f_means) / f_std #матрица
    return f_stand
```

_Стандартизируем признаки._

```python
X_stand = standart(X)
#также добавим свободный коэффициент обратно
X_stand = np.c_[X_stand, np.ones(X_stand.shape[0])]
X_stand
```

_Снова запустим обучение. На обновленных признаках._

```python
epohs = 10000
for _ in range(epohs):

    w = grad(X_stand, w, Y, lr)
    pred = X_stand @ w
print(w.round(1))
```
