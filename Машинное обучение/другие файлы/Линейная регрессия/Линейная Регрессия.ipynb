```python
import numpy as np
import seaborn as sns
```

```python
x = np.random.randint(1, 100, 50)
```

```python
y = 2*x
nois = np.random.normal(0, 10, size=y.shape)
y = y + nois
y1 = 2*x
y2 = 1.5*x - 30
y3 = 2.5*x
y4 = 1.2*x + 50
sns.scatterplot(x = x, y = y)
sns.lineplot(x = x, y = y1, color='#f00', label='y1 = 2*x')
# sns.lineplot(x = x, y = y2, color='#0ff', label='y2 = 1.5*x - 30')
# sns.lineplot(x = x, y = y3, color='#0f0', label='y3 = 2.5*x')
# sns.lineplot(x = x, y = y4, color='#990', label='y4 = 1.2*x + 50')
```

```python
correlation = np.corrcoef(x, y)[0,1]
print(f"Корреляция между x и y: {correlation:.2f}")
```

```python
correlation = np.round(np.corrcoef(x, y), 2)
print(f"Тепловая Матрица: \n {correlation}")
sns.heatmap(correlation)
```
