```python
import numpy as np
x1 = np.array([2, 13, 8, 5, 9])
x2 = np.array([8, 15, 24, 18, 7])
y = 8 * x1 + 14 * x2
```

```python
x1, x2, y
```

```python
mx = np.column_stack((x1, x2))
mx
```

```python
w = np.array([1, 1])
lr = 1e-3
ep = 1000
for _ in range(ep):
    pr = mx @ w.T
    er = (pr - y)[:, None]
    gr = np.mean(2 * er * mx, axis=0)
    w = w - lr * gr
    print(w)
w
```
