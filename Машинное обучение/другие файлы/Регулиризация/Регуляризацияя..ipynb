```python
import numpy as np
import pandas as pd

from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression, Ridge
from sklearn.metrics import mean_squared_error


# ============================================================
# 1. СОЗДАЁМ ДАННЫЕ
# ============================================================

np.random.seed(33)

n = 300

# Настоящий скрытый фактор
x = np.random.uniform(1, 100, n)


# Создаём признаки с сильной мультиколлинеарностью
X = pd.DataFrame()

X["f0"] = x

for i in range(1, 20):
    X[f"f{i}"] = x + np.random.normal(0, 0.01, n)


# Целевая переменная
# На самом деле y зависит только от x
y = 5 * x + np.random.normal(0, 20, n)


print("Размер X:", X.shape)
print("Размер y:", y.shape)


# ============================================================
# 2. ПРОВЕРЯЕМ МУЛЬТИКОЛЛИНЕАРНОСТЬ
# ============================================================

print("\nКорреляция признаков:")
print(X.corr().round(3))


# Проверяем ранг матрицы
rank = np.linalg.matrix_rank(X)

print("\nКоличество признаков:", X.shape[1])
print("Ранг матрицы X:", rank)


# ============================================================
# 3. ДЕЛИМ ДАННЫЕ НА TRAIN / TEST
# ============================================================

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.3,
    random_state=33
)


# ============================================================
# 4. LINEAR REGRESSION БЕЗ РЕГУЛЯРИЗАЦИИ
# ============================================================

linear_model = LinearRegression()

linear_model.fit(X_train, y_train)

y_train_pred = linear_model.predict(X_train)
y_test_pred = linear_model.predict(X_test)


mse_train_linear = mean_squared_error(
    y_train,
    y_train_pred
)

mse_test_linear = mean_squared_error(
    y_test,
    y_test_pred
)


print("\n==============================")
print("LINEAR REGRESSION")
print("==============================")

print("MSE train:", mse_train_linear)
print("MSE test :", mse_test_linear)

print("\nКоэффициенты:")
print(linear_model.coef_)


# ============================================================
# 5. RIDGE — РЕГУЛЯРИЗАЦИЯ
# ============================================================

ridge_model = Ridge(alpha=10)

ridge_model.fit(X_train, y_train)

y_train_pred_ridge = ridge_model.predict(X_train)
y_test_pred_ridge = ridge_model.predict(X_test)


mse_train_ridge = mean_squared_error(
    y_train,
    y_train_pred_ridge
)

mse_test_ridge = mean_squared_error(
    y_test,
    y_test_pred_ridge
)


print("\n==============================")
print("RIDGE")
print("==============================")

print("MSE train:", mse_train_ridge)
print("MSE test :", mse_test_ridge)

print("\nКоэффициенты:")
print(ridge_model.coef_)


# ============================================================
# 6. СРАВНИВАЕМ РЕЗУЛЬТАТЫ
# ============================================================

print("\n==============================")
print("СРАВНЕНИЕ")
print("==============================")

print(
    f"LinearRegression MSE: {mse_test_linear:.2f}"
)

print(
    f"Ridge MSE:            {mse_test_ridge:.2f}"
)


# ============================================================
# 7. СМОТРИМ, СТАЛО ЛИ ЛУЧШЕ
# ============================================================

improvement = (
    (mse_test_linear - mse_test_ridge)
    / mse_test_linear
    * 100
)

print(
    f"\nУлучшение: {improvement:.2f}%"
)


# ============================================================
# 8. СРАВНИВАЕМ КОЭФФИЦИЕНТЫ
# ============================================================

coefficients = pd.DataFrame({
    "LinearRegression": linear_model.coef_,
    "Ridge": ridge_model.coef_
}, index=X.columns)

print("\n==============================")
print("КОЭФФИЦИЕНТЫ")
print("==============================")

print(coefficients)
```
