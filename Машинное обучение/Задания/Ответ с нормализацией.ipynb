```python
import numpy as np
np.random.seed(1)
ws = np.array([8, 15, 22])
x = np.random.randint(30, 100, size=(50, 3))
y = x @ ws + 300

x_t = x[25:]
x = x[:25]

y_a = (y + np.random.normal(0, 20, size=y.shape)).astype(int)
y = y_a[:25]
y_t = y_a[25:]
mt_train = np.column_stack((x, y))
mt_test = np.column_stack((x_t, y_t))
np.savetxt('mt_train.csv', mt_train, delimiter=';', fmt='%d')
np.savetxt('mt_test.csv', mt_test, delimiter=';', fmt='%d')


# Нормализация обучающих данных (по столбцам)
x_mean = x[:, :3].mean(axis=0)
x_std = x[:, :3].std(axis=0) + 1e-8  # +1e-8 чтобы не делить на 0
x_norm = (x[:, :3] - x_mean) / x_std

# Добавляем колонку единиц уже к нормализованным данным
x = np.column_stack((x_norm, np.ones(x.shape[0])))
```

```python
w = np.array([0, 0, 0, 0], dtype=float)
lr = 1e-2
rmse_lst = 1e+10
for ep in range(1000):
    pr = x @ w
    er = pr - y
    gr = np.mean(2 * er[:, None] * x, axis=0)
    w = w - lr * gr
    rmse = np.sqrt(np.mean((pr - y)**2))
    if rmse_lst - rmse < 0.001:
        print(rmse_lst, rmse)
        break
    else:
        rmse_lst = rmse
    if ep % 100 == 0:
        print(ep, w, rmse)
print(w, rmse)
```

```python
def pred_er(y, x, w):
    rmse = np.sqrt(np.mean(((x @ w.T) - y)**2))
    return rmse
print(pred_er(y_t, x_t, w))
```
